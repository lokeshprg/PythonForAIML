## Disadvantages

- Performance: Python is an interpreted language, which means that it can be slower than compiled languages like C or Java. This can be an issue for performance-intensive tasks.
- Global Interpreter Lock: The Global Interpreter Lock (GIL) is a mechanism in Python that prevents multiple threads from executing Python code at once. This can limit the parallelism and concurrency of some applications.
- Memory Consumption: Consumes a lot of memory when working with large datasets and or running complex algorithms
- Dynamically typed: Dynamically typed language, which means that the types of variables can change at runtime. This can make it more difficult to catch errors and can lead to bugs.
- Packaging and versioning: It has a large number of packages and libraries, which can sometimes lead to versioning issues and package conflicts.


# Input & Output in Python

**Taking Input using Input()**<br>
input() function is used to take user input. By default, it returns user input in form of string.

```python
name = input("Enter Your Name: ")
print("Hello", name, "! Welcome!")
```

<hr>


**Printing output using output()**<br>
print() function allows us to display text, variables and expressions on the console. In the example below string literal is printed in the console using print() function by enclosing text in ' or " quotes.

```python
print("Hello World!")
print('Hello World!')
```

**Printing Variables using print()**<br>
We can print single or multiple variables using print() function.

```python
s = "Hello"
print(s)

a = "Alice"
b = 36
c = "New York"
print(a, b, c)
```

# Concept of Object Reference
let us assign a variable x to value 5.<br>
***x=5***<br>
When x = 5 is executed, Python creates an object to represent the value 5 and makes x reference this object.<br>
![x=5](images/1.WEBP)<br>
Now, lets assign another varible y to variable x.<br>
***y=x***<br>
This statement creates y reference the same object as x, not x itself.This is called *Shared Reference*.<br>
![y=x](images/2.WEBP)<br>
now, if we write<br>
***x = "Geeks"***<br>
Python creates a new object for value "Geeks" and makes x reference this new object.<br>
![x="Geeks"](images/3.WEBP)<br>
The varible y remain unchanged, still referencing to the original object 5. now, if we assign a new value to y:<br>
***y="Computer"***<br>
![y="Computer"](images/4.WEBP)<br>
Python variables store references to objects, not the actual values themselves. When a variable is reassigned, it starts referencing a new object while the old unreferenced object becomes eligible for garbage collection.<br>
**Understanding Variable Reassignment**<br>
In this example, We will understand whether the value of an variable changes when changing another variable, initially both of them referencing to the same object.

```python
x = 1
y = x
y = y+1
print(x, y)
```

In the above example, initially both the objects reference to object 1.<br>
After y = y+1, references y to new object 2 while x still references 1, so changing y does not affect x.

# Python Operators
In python programming, Operators are used to perform operations on the Operands/Values/Variables.
- Operators: Special Symbols like +, *, =, /, etc.
- Operands: Values on which operation is applied.<br><br>
![Operators](images/pythiob-operators.webp)<br>
## Arithmetic Operator
In python, arithmetic operators are used to used to perform basic operations like addition, subtraction, multiplication and division.
- /(division operator): Returns floating point number.
- //(floor division): Returns integer result.

```python
a = 15
b = 4
print("Addition:",a+b)
print("Subtraction:", a-b)
print("Multiplication:", a*b)
print("Division:", a/b)
print("Floor Division:",a//b)
print("Modulus:",a%b)
print("Exponentiation:",a**b)
```

## Conditional Operator or Relational Operator
Conditional (or Relational) operators are used to compare values. These operators either return True or False based on the condition.

```python
a = 13
b = 33

print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a >= b)
print(a <= b)
```

## Logical Operators
Logical operators are used to perform LOGICAL AND, LOGICAL AND, and LOGICAL OR. It is used to to combine conditional statements.
**Precedence of Logical Operators is:**
- LOGICAL NOT
- LOGICAL AND
- LOGICAL OR

```python
a = True
b = False
print(a and b)
print(a or b)
print(not a)
```

## Bitwise Operators
Bitwise operators acts on bits and perform bit-by-bit operations. These are used to operate on binary numbers.
Following are the Bitwise Operators:
- BITWISE NOR
- BITWISE OR
- BITWISE XOR
- BITWISE AND
- BITWISE SHIFT

```python
a = 10
b = 4
print(a&b)
print(a|b)
print(a^b)
print(a>>2)
print(a<<2)
```
