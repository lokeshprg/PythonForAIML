# Concept of Object Reference
let us assign a variable x to value 5.<br>
***x=5***<br>
When x = 5 is executed, Python creates an object to represent the value 5 and makes x reference this object.<br>
![x=5](images/1.WEBP)<br>
Now, lets assign another varible y to variable x.<br>
***y=x***<br>
This statement creates y reference the same object as x, not x itself.This is called *Shared Reference*.<br>
![y=x](images/2.WEBP)<br>
now, if we write<br>
***x = "Geeks"***<br>
Python creates a new object for value "Geeks" and makes x reference this new object.<br>
![x="Geeks"](images/3.WEBP)<br>
The varible y remain unchanged, still referencing to the original object 5. now, if we assign a new value to y:<br>
***y="Computer"***<br>
![y="Computer"](images/4.WEBP)<br>
Python variables store references to objects, not the actual values themselves. When a variable is reassigned, it starts referencing a new object while the old unreferenced object becomes eligible for garbage collection.<br>
**Understanding Variable Reassignment**<br>
In this example, We will understand whether the value of an variable changes when changing another variable, initially both of them referencing to the same object.

```python
x = 1
y = x
y = y+1
print(x, y)
```

In the above example, initially both the objects reference to object 1.<br>
After y = y+1, references y to new object 2 while x still references 1, so changing y does not affect x.

# Python Data Types
Data Types in python define the type of value stored in the variable and determine the operations that can be performed on that data. Since python treats everything as object, each value is assigned with a specific data type.
- It helps the interpreter to store and process different kind of data more efficiently.
- Enables correct operations by ensuring valid actions are performed on compatible data types.<br><br>

## Numeric Data Types
Numeric data types are used to store numeric values. It can be an integer, floating number or even a complex number. Python supports threee main numeric types:
- Integers: Reperesented by int class. It is used to store positive and negative whole numbers (without fractions or decimals).
- Float: Represented by Float class. Used to store real numbers with floating point representation. It is specified by a decimal point.
- Complex: It is represented by complex class. It is used to store number with real and imaginary parts. For Example: 2+4j.

```python
a = 3
b = 5.6
c = 9+7j
print(type(a))
print(type(b))
print(type(c))
```

## Sequence Data Types
Sequence data types is collection of ordered items, that can be of similar or different data types. Elements in sequence can be accessed using indexing.<br>
***1. String***<br>
String are used to store text data. It can be represented using single, double or triple quote. A string is represented using str class. 

```python
a = "Hello World"
print(a)
print(type(a))
# Accessing the string with index
print("First index is:", a[0])
print("Last index is:", a[-1])
```

## Lists
Lists are ordered and mutable collection of data. It stores multiple values in single variable which can be of same or different data types. elements in a list can be accessed using indexing.

```python
a = [1, 45, "Hello", "Python"]
print(a)
print(type(a))
# Accessing List Items
print(a[2])
print(a[-1])
```

## Tuples
Tuples are ordered and immutable collection of data used to store multiple values in single variable. once tuple is created, tuple elements cannot be modified and are accessed using indexing.<br>
*Note: Tuple with single elemnet must include a trailing comma, otherwise python treat it as normal value instead of tuple.*<br>

```python
a = ("hi",)
print(a)
print(type(a))

b = (1, 56, "Hello", "World")
print(b)
print(type(b))
# Accessing tuple Items
print(b[1])
print(b[-2])
```

# Boolean Data Type
Boolean data type represents one of the two values: True or False. It is mainly used in conditions and comparisions, it represented by bool class.

## Set Data Type
Set data type is unordered and mutable collection of unique elements. Since, sets are unordered we cannot access the elements in sets using indexing. To access the elements of sets we have to iterate through each element using a loop.<br>
*Note: sets are unordered, so the order of elements may vary each time when printed on running the code.*

```python
a = {"a", "b", "c", "a", "b"}
print(a)

b = {"Hello", "World", "Python"}
for i in b:
    print(i)
```

## Dictionary Data Type
Dictionaries are used to data in key:value pairs. Each key in a dictionary must be unique and value are accessed using their keys with square brackets [] or **get()** method. Dictionary keys are case-sensitive.

# Conditional Statements in Python
Conditional statement in python are used to control the flow of program based on specific conditions. They allow a program to execute different block of code depending on whether a condition is True or False.
## if Statement
If statement is used to execute the block of code only when a specified condition evaluates to True.<br><br>
![if](images/if.webp)

```python
age = 18
if age >= 18:
    print("You are Eligible to Vote")
```

## Short Hand if
Short hand if is used to write if statements in a single line. It is useful only when one statement needs to be executed.

```python
age = 19
if age > 18: print("Eligible to vote")
```

## if else Statement
If else statement is used to execute one block of code when the condition is True and another block when the condition is False.<br><br>
![ifelse](images/ifelse.webp)

```python
age = 10
if age <= 12:
    print('Travel for free.')
else:
    print('Pay for Ticket.')
```

## if-elif-else Statement
elif statement is used to check multiple conditions. It is executes the block of code only when its condition is True and the previous condition is evaluated to False.

```python
age = 25
if age <= 12:
    print('Child')
elif age <= 19:
    print('Teenager')
elif age <= 35:
    print('Young Adult')
else:
    print('Adult')
```

## Nested if-else Statement
Nested if-else statement the if-else statement placed inside another if or else block. It is used when the condition needs to be checked within another condition.


```python
age = 70
is_Member = True

if age >= 60:
    if is_Member:
        print('30% Discount  is applied')
    else:
        print('20% Dicount id applied')
else:
    print('You are not eligible for the Senior Discount')
```

## Conditional Expression (Ternary Operator)
COnditional Statement or Ternary operator is used when if-else statement is used in the single line. It is used when choosing between two values based on a condition

```python
age = 20
s = 'Adult' if age >= 18 else 'Child'
print(s)
```

## Match-Case Statement
Match-Case statement is used to compare value with multiple patterns and execute the matching block of code. It is similar to the switch-case statement availaible in other programming language.

```python
number = 2
match number:
    case 1:
        print('one')
    case 2|3:
        print('Two or Three')
    case _:
        print('Other Number')
```

# Loops in Python
Loops are used to execute the block of code repeatedly until a condition is met or all items in a sequence are processed. The main types of loop are For Loops (Iterating over a sequence) and While loop (executing code based on the condition).
## For Loop
For loops are used to iterate through each element in a sequence like list, tuple, string or range. It executes once for all items in sequence

```python
n = 4
for i in range(0, n):
    print(i)
```

![for loop](images/forloop.webp)

***Iterating by index of Sequences***<br>
A For loop can also iterate through the sequence elements by using their index values.


```python
lst = ['Geeks', 'For', 'Geeks']
for i in range(len(lst)):
    print(lst[i])
```

## While Loops
While loops repeatedly executes the block of code till the given condition evaluates to True. When the given condition becomes False, It executes the next line code after the loop immediately.

```python
n = 0
while (n<3):
    print('Hello World!')
    n+=1
```

**Explanation**:<br>
- n=0: initializes the counter variable
- while (n<3): Executes the block of code until the condition is true
- n+=1: increases the counter value by 1 <br>
![while loop](images/whileloop.webp)

## Nested Loop
Nested loop is the loop inside another loop. The inner loop executes completely for each iteration of the outer loop.

```python
for i in range(1, 5):
    for j in range(i):
        print(i, end=' ')
    print()
```

**Explanation:**

- Outer loop for i in range(1, 5) controls the number of rows and Inner loop for j in range(i) runs i times for each row.
- print(i, end=' ') prints the value of i on the same line and print() moves output to the next line after each row.
